### 1. Giới thiệu Bối Cảnh và Đóng Góp Khoa Học Của Bài Báo
Bài báo khảo sát các phương pháp phòng thủ cơ sở (Baseline Defenses) chống lại các cuộc tấn công đối kháng sinh hậu tố độc hại (như GCG - Zou et al., 2023). Tác giả phát hiện ra rằng các đòn tấn công đối kháng tối ưu hóa gradient tạo ra các chuỗi ký tự bất thường, có thể bị phát hiện hiệu quả bằng **lọc Perplexity** và **phân loại n-grams ký tự (Character N-Grams)** với chi phí tính toán cực kỳ thấp trên CPU mà không cần chạy mô hình lớn.

### 2. Kiểm Tra Tập Dữ Liệu Khép Kín Cục Bộ (`datasets/jain_eval_benchmark.json`)

```python
import os
import json

dataset_file = os.path.join('datasets', 'jain_eval_benchmark.json')
with open(dataset_file, 'r', encoding='utf-8') as f:
    dataset = json.load(f)

n_benign = sum(1 for x in dataset if x['label'] == 0)
n_attack = sum(1 for x in dataset if x['label'] == 1)
print(f'Tổng số mẫu: {len(dataset)}')
print(f'  - Mẫu Lành Tính (Benign - Nhãn 0): {n_benign}')
print(f'  - Mẫu Tấn Công (GCG / Jailbreak - Nhãn 1): {n_attack}')
print(f'Ví dụ mẫu tấn công [0]:\n  Prompt: {dataset[0]["prompt"][:120]}...\n  Source: {dataset[0]["source"]}')
```
